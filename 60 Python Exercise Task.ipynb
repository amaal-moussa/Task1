**22-Given a list in Python and provided the positions of the elements, write a program to swap the two elements in the list. (Input : List = [23, 65, 19, 90], pos1 = 1, pos2 = 3
Output : [19, 65, 23, 90])**

```python
def swap_elements(lst, pos1, pos2):
    pos1 -= 1
    pos2 -= 1
    
    lst[pos1], lst[pos2] = lst[pos2], lst[pos1]
    return lst

my_list = [23, 65, 19, 90]
pos1 = 1
pos2 = 3

result = swap_elements(my_list, pos1, pos2)
print(result)
```

**29-Given list values and keys list, convert these values to key value pairs in form of list of dictionaries. (Input : test_list = [“Gfg”, 3, “is”, 8], key_list = [“name”, “id”]
Output : [{‘name’: ‘Gfg’, ‘id’: 3}, {‘name’: ‘is’, ‘id’: 8}])**

```python
test_list = ["Gfg", 3, "is", 8]
key_list = ["name", "id"]
result = []
step = len(key_list)

for i in range(0, len(test_list), step):
    values = test_list[i:i+step]
    result.append(dict(zip(key_list, values)))

print(result)
```

**35-write a Python program to get all unique combinations of two Lists (List_1 = ["a","b"]
List_2 = [1,2]
Unique_combination = [[('a',1),('b',2)],[('a',2),('b',1)]] )**

```python
List_1 = ["a", "b"]
List_2 = [1, 2]

result = []

for i in range(len(List_2)):
    for j in range(len(List_2)):
        if i != j:
            perm = [List_2[i], List_2[j]]
            
            pair = []
            for k in range(len(List_1)):
                pair.append((List_1[k], perm[k]))
            
            result.append(pair)

print(result)
```

```python
test_dict = {'Gfg': 3, 'is': 7, 'best': 10, 'for': 6, 'geeks': 'CS'}
K = 7

new_dict = {key: value for key, value in test_dict.items()
            if not (isinstance(value, (int, float)) and value > K)}

print(new_dict)
```
